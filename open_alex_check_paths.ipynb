```python
import json
import os
import pickle as pkl
import json
```

```python
open_alex_dir = "open_alex_citations"
not_found = []
found = []
found_count= 0
files = os.listdir(open_alex_dir)
for file in files:
    file_path = os.path.join(open_alex_dir, file)
    
    with open(file_path, 'rb') as f:
            pmids = json.load(f)
    if "not_found" in file:
        not_found.extend(pmids)
    else:
        found.extend(pmids)
        found_count+=1
```
