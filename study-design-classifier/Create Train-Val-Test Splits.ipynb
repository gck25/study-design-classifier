```python
import pickle as pkl
from sklearn.model_selection import train_test_split
from proposed_design_types import PROPOSED_DESIGN_TYPES
```

```python
ids_file = '/home/k2141255/study_design_classifier/data/cleaned_pos_neg_dataset_ids.pkl'
with open(ids_file, 'rb') as f:
    ids = pkl.load(f)
```

```python
random_seed = 42
test_size = 0.1
val_size = 0.1
```

```python
ids_split = {}
for sdtype in ids.keys():
#     if sdtype in PROPOSED_DESIGN_TYPES:
        pos_ids = ids[sdtype]['pos']
        neg_ids = ids[sdtype]['neg']
        
        train_val_pos, test_pos = train_test_split(pos_ids, test_size=test_size, random_state=random_seed, shuffle=True)
        train_pos, val_pos = train_test_split(train_val_pos, test_size=val_size, random_state=random_seed, shuffle=True)
        
        train_val_neg, test_neg = train_test_split(neg_ids, test_size=test_size, random_state=random_seed, shuffle=True)
        train_neg, val_neg = train_test_split(train_val_neg, test_size=val_size, random_state=random_seed, shuffle=True)
        
        print(sdtype)
        print('pos')
        print('train:',len(train_pos))
        print('val:', len(val_pos))
        print('test:', len(test_pos))
        
        print('neg')
        print('train:', len(train_neg))
        print('val:', len(val_neg))
        print('test:', len(test_neg))
        
        ids_split[sdtype] = {'pos':
                             {
                                 'train': train_pos,
                                 'val': val_pos,
                                 'test': test_pos
                             },
                             'neg':
                             {
                                 'train': train_neg,
                                 'val': val_neg,
                                 'test': test_neg
                             }
                            }
```

```python
ids_train_val_test_file = '/home/k2141255/study_design_classifier/data/cleaned_pos_neg_train_val_test_dataset_ids.pkl'
```

```python
with open(ids_train_val_test_file, 'wb') as f:
    pkl.dump(ids_split, f)
```
