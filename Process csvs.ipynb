```python
import csv
import sys
import pandas as pd
import random
from proposed_design_types import PROPOSED_DESIGN_TYPES
filename = 'pubmed_articles_pos.csv'
```

```python
with open(filename, newline='') as csvfile:
    spamreader = csv.reader(csvfile, delimiter=' ', quotechar='|')
    counter = 0
    for row in spamreader:
        print(', '.join(row))
        if counter ==5:
            break
        counter +=1
```

```python

chunksize = 2000
df_iter = pd.read_csv(filename, chunksize=chunksize)
```

```python
df_iter
```

```python
df_lst = [] 

# Iterate over the file based on the criteria and append to the list
for df_ in df_iter: 
        tmp_df = (df_.rename(columns={col: col.lower() for col in df_.columns}) # filter eg. rows where 'col_1' value grater than one
                                  .pipe(lambda x:  x[x['indexing_method'] == 'Human'] ))
        df_lst += [tmp_df.copy()] 

# And finally combine filtered df_lst into the final lareger output say 'df_final' dataframe 
df_final = pd.concat(df_lst)
```

```python
df_final
```

```python
for des_typ in PROPOSED_DESIGN_TYPES:
    print(des_typ, len(df_final[[des_typ in typs for typs in df_final['ptyp']]]))
```

```python
with open('pubmed_articles_pos_human.csv', 'w') as f_test:
    df_final.to_csv(f_test)
```
