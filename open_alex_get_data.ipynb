```python
import json
import requests
import pickle as pkl
from tqdm import tqdm
import pandas as pd
import sys
import time
import os
import random
from random import choices
from numba import jit, cuda
from gpuparallel import GPUParallel, delayed
import itertools
```

```python
with open('../data/cleaned_pos_neg_train_val_test_dataset_ids.pkl', 'rb') as f:
    prop_typ_id_dict= pkl.load(f)
```

```python
prop_typ_id_dict.keys()
```

```python
pos_indices2pmid = {}
for study in prop_typ_id_dict.keys():
    study_dict = prop_typ_id_dict[study]
    for pos_neg in study_dict.keys():
        pos_neg_dict = study_dict[pos_neg]
        for split in pos_neg_dict:
            for ind in pos_neg_dict[split]:
                if pos_neg == 'pos':
                    pos_indices2pmid[ind] = []
#             indices.extend(list(pos_neg_dict[split]))
```

```python
neg_indices2pmid = {}
for study in prop_typ_id_dict.keys():
    study_dict = prop_typ_id_dict[study]
    for pos_neg in study_dict.keys():
        pos_neg_dict = study_dict[pos_neg]
        for split in pos_neg_dict:
            for ind in pos_neg_dict[split]:
                if pos_neg == 'neg':
                    neg_indices2pmid[ind] = []
```

```python
row_ids_seen = []
```

```python
filename_neg_human = '../data/pubmed_articles_neg_human.csv'
neg_pmids = []

chunksize = 10**4
for chunk in tqdm(pd.read_csv(filename_neg_human, chunksize=chunksize)):
    row_ids = list(set(neg_indices2pmid.keys()).intersection(set(chunk.index)))
    row_ids_seen.extend(row_ids)
    for row_id in row_ids:
        neg_indices2pmid[row_id].append(chunk.loc[row_id]['pmid'])
#     neg_pmids.extend(chunk.loc[row_ids]['pmid'])
```

```python
filename_pos_human = "../data/pubmed_articles_pos_human.csv"
pos_pmids = []

chunksize = 10**4
for chunk in tqdm(pd.read_csv(filename_pos_human, chunksize=chunksize)):
    row_ids = list(set(pos_indices2pmid.keys()).intersection(set(chunk.index)))
    row_ids_seen.extend(row_ids)
    for row_id in row_ids:
        pos_indices2pmid[row_id].append(chunk.loc[row_id]['pmid'])
#     pos_pmids.extend(chunk.loc[row_ids]['pmid'])
```
