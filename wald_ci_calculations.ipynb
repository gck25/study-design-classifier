```python
def wilson_confidence_interval(p, n, c):
    
    p_hat = np.sqrt(p**2 + p*(1-p)/n)
    
    z_score = st.norm.ppf(1 - ((1 - c) / 2))

    interval = z_score * np.sqrt(p_hat * (1 - p_hat) / n)
    
    return interval
```

```python
def adjust_prob(p, n):
    return np.sqrt(p**2 + p*(1-p)/n)
```

```python
def wilson_confidence_interval_precision(p, R, S, dR, dS, tn, tp, c):
    
    p_hat = adjust_prob(p, tp)
    R_hat = adjust_prob(R, tp)
    S_hat = adjust_prob(S, tn)
    
    interval = np.sqrt((R_hat**2 * (dR + (tn/tp*dS))**2)/
                      (R + tn/tp * (1 - S_hat)) ** 4
                      + (dR / (R + tn/tp * (1 - S_hat))) **2 )
    
#     interval = p * np.sqrt((dR/R)**2 + (dR**2 + (tn/tp*dS)**2) /
#                            (R + tn/tp * (1-S))**2 )
    
    return interval
```
