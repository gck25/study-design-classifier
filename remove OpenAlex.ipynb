```python
import pickle as pkl
import os
from tqdm import tqdm
from joblib import Parallel, delayed
```

```python
with open('../data/cleaned_pos_neg_train_val_test_dataset_ids.pkl', 'rb') as f:
    prop_typ_id_dict= pkl.load(f)
```

```python
pmids = []
for sd in prop_typ_id_dict.keys():
    pmids_sd = prop_typ_id_dict[sd]
    for pn in pmids_sd.keys():
        pmids_pn = pmids_sd[pn]
        for dset in pmids_pn.keys():
            pmids.extend(pmids_pn[dset])
```
