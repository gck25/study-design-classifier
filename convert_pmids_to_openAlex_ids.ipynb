```python
import json
import requests
import pickle as pkl
from tqdm import tqdm
```

```python
with open("../data/cleaned_pos_neg_train_val_test_dataset_ids.pkl", "rb") as f:
    prop_typ_id_dict = pkl.load(f)
```

```python
PMIDS = '../data/pmids.pkl'
with open(PMIDS, 'rb') as f:
    pmids = pkl.load(f)
```

```python
pmids = []
for split in prop_typ_id_dict:
    for sd in prop_typ_id_dict[split]:
        for pos_neg in prop_typ_id_dict[split][sd]:
            pmids.extend(prop_typ_id_dict[split][sd][pos_neg])
```

```python
pmid_to_openAlex = {}
```

```python
batch_size = 50
num_batches = len(pmids)//batch_size

for i in tqdm(range(num_batches)):
    pmid_str = '|'.join(map(str, pmids[i*batch_size: (i+1)*batch_size]))
    response = requests.get(f"https://api.openalex.org/works?filter=pmid:{pmid_str}&per-page=200")
    while response.status_code != 200:
        response = requests.get(f"https://api.openalex.org/works?filter=pmid:{pmid_str}&per-page=200")
    
```
